Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Régression linéaire** (*Régression linéaire*) appartient à la famille de **régression**. Son objectif est de prédire une variable quantitative à partir d'une ou plusieurs variables explicatives. L'intuition est de rechercher la relation linéaire qui minimise la somme des carrés des erreurs. Le fonctionnement suit l'enchaînement suivant : **X,y → estimation des coefficients → prédictions ŷ → mesure de l'erreur.** La relation ou le critère essentiel est : ŷᵢ = β₀ + Σⱼ βⱼxᵢⱼ ; β₀ est l'interception, βⱼ le coefficient appris pour la variable xⱼ, ŷᵢ la prédiction et xᵢⱼ la valeur de la variable j pour l'observation i. Les β sont des paramètres appris ; il n'y a pas d'hyperparamètre dans l'appel LinearRegression() utilisé ici. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** simple et rapide ; coefficients relativement faciles à interpréter. **Limites :** relation linéaire nécessaire pour de bonnes performances ; sensible aux valeurs aberrantes et à la multicolinéarité.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Ridge** (*Ridge*) appartient à la famille de **régression**. Son objectif est de réduire le surapprentissage et la sensibilité à la multicolinéarité par une régularisation L2. L'intuition est de pénaliser les coefficients trop grands pour obtenir un modèle plus stable. Le fonctionnement suit l'enchaînement suivant : **standardisation → apprentissage pénalisé → prédiction.** La relation ou le critère essentiel est : minβ Σᵢ(yᵢ-β₀-xᵢᵀβ)² + αΣⱼβⱼ² ; α est l'hyperparamètre de régularisation et β les paramètres appris. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** plus stable en présence de variables corrélées ; réduit souvent le surapprentissage. **Limites :** les coefficients ne deviennent généralement pas exactement nuls ; le choix de α influence fortement le résultat.

### Présentation

L'intuition est de transformer les variables afin qu'une régression linéaire puisse représenter une courbure. Le fonctionnement suit l'enchaînement suivant : **création des caractéristiques polynomiales → standardisation → Ridge → prédiction.** La relation ou le critère essentiel est : ŷ = β₀ + Σβⱼzⱼ, où z contient les variables originales, leurs puissances et interactions générées par PolynomialFeatures. degree=2 est l'hyperparamètre de degré. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** capture des relations courbes simples ; reste relativement interprétable pour un petit nombre de variables. **Limites :** le nombre de variables peut exploser avec le degré ; risque de surapprentissage pour des degrés élevés.

Le fonctionnement suit l'enchaînement suivant : **recherche d'une coupure → séparation → répétition → prédiction dans une feuille.** La relation ou le critère essentiel est : À chaque nœud, on cherche une coupure réduisant l'impureté de régression, souvent liée à l'erreur quadratique ; une feuille produit une valeur prédite, typiquement une moyenne des observations qui y arrivent. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** capture les non-linéarités et interactions ; peu de prétraitement nécessaire. **Limites :** risque de surapprentissage ; instable : de petites variations des données peuvent modifier l'arbre.

### Présentation

**Bandits manchots** (*Bandits manchots*) appartient à la famille de **renforcement**. Son objectif est de choisir parmi plusieurs actions dont les récompenses sont inconnues afin de maximiser la récompense cumulée. L'intuition est de apprendre quel bras rapporte le plus en essayant les différentes options. Le fonctionnement suit l'enchaînement suivant : **choisir un bras → recevoir une récompense → mettre à jour l'estimation → recommencer.** La relation ou le critère essentiel est : La valeur estimée d'un bras a peut être Q(a)=Σ récompenses / nombre de choix. Le choix de l'action est une décision ; les probabilités de récompense sont ici définies dans la simulation. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** problème simple pour étudier exploration/exploitation ; utile pour recommandation et allocation en ligne. **Limites :** environnement simplifié ; ne modélise pas directement un état complexe comme un MDP.

Le fonctionnement suit l'enchaînement suivant : **tirage aléatoire → exploration si <ε → sinon meilleure action estimée → récompense → mise à jour.** La relation ou le critère essentiel est : avec probabilité ε, choisir une action exploratoire ; sinon choisir argmaxₐ Q(a). ε=0.1 est l'hyperparamètre d'exploration dans l'exemple. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** très simple ; permet un contrôle direct de l'exploration. **Limites :** ε fixe peut être trop élevé ou trop faible ; ne garantit pas une exploration optimale.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Présentation

**Q-learning** (*Q-learning*) appartient à la famille de **renforcement**. Son objectif est de apprendre la valeur d'une action dans un état afin de choisir une politique maximisant les récompenses futures. L'intuition est de mettre à jour la valeur d'une action en fonction de la récompense reçue et de la meilleure valeur future estimée. Le fonctionnement suit l'enchaînement suivant : **observer s → choisir a → recevoir r et s' → mise à jour Q → recommencer.** La relation ou le critère essentiel est : Q(s,a)←Q(s,a)+α[r+γ maxₐ'Q(s',a')−Q(s,a)], où α est le taux d'apprentissage, γ le facteur d'actualisation, r la récompense, s l'état et s' le nouvel état. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** simple pour petits espaces d'états discrets ; apprend une politique sans modèle explicite de l'environnement. **Limites :** table Q impossible à grande dimension ; sensibilité à α, γ et à la stratégie d'exploration.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Présentation

**SARSA** (*SARSA*) appartient à la famille de **renforcement**. Son objectif est de apprendre une fonction Q selon les actions effectivement suivies, en tenant compte de la politique d'exploration. L'intuition est de contrairement à Q-learning, la mise à jour utilise la prochaine action effectivement sélectionnée par la politique. Le fonctionnement suit l'enchaînement suivant : **s,a → interaction → r,s' → choix de a' → mise à jour Q(s,a) → continuer.** La relation ou le critère essentiel est : Q(s,a)←Q(s,a)+α[r+γQ(s',a')−Q(s,a)], où a' est l'action réellement choisie dans le nouvel état. Les paramètres sont appris pendant l'entraînement, tandis que les hyperparamètres sont fixés avant l'apprentissage. **Avantages :** apprentissage en ligne simple ; prend directement en compte la politique suivie. **Limites :** performance dépendante de la stratégie d'exploration ; table limitée aux petits espaces d'états.

### Présentation

Son objectif est de approximer la fonction Q par un réseau neuronal lorsque l'espace d'états est trop grand pour une table. L'intuition est de remplacer la table Q par un réseau neuronal capable d'approximer les valeurs d'actions. Le fonctionnement suit l'enchaînement suivant : **état → réseau neuronal → valeurs Q → action → récompense → mémoire de rejeu → mise à jour du réseau.** La relation ou le critère essentiel est : y=r+γ maxₐ' Q_target(s',a') si l'état n'est pas terminal ; la perte compare Q_online(s,a) à cette cible de Bellman.
